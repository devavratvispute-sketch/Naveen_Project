```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from datetime import datetime
```

```python
def current_time():
    return current_timestamp()
```

```python
def add_audit_columns(df, source_name):
    return (
        df.withColumn("processed_date", current_timestamp())
          .withColumn("load_date", current_date())
          .withColumn("source_system", lit(source_name))
    )
```

```python
def read_table(table_name):
    return spark.table(table_name)
```

```python
def write_delta(df, table_name, mode="overwrite"):
    (
        df.write
          .format("delta")
          .mode(mode)
          .option("mergeSchema", "true")
          .saveAsTable(table_name)
    )
```

```python
def rename_columns(df):
    for col_name in df.columns:
        df = df.withColumnRenamed(col_name, col_name.strip().replace(" ", "_").lower())
    return df
```

```python
def remove_duplicates(df):
    return df.dropDuplicates()
```

```python
def replace_nulls(df):
    return df.fillna("NA")
```

```python
def notebook_start_time():
    start_time = datetime.now()
    print("Notebook Started :", start_time)
    return start_time
```

```python
def notebook_end_time():
    end_time = datetime.now()
    print("Notebook Ended :", end_time)
    return end_time
```

```python
def time_diff(start_time, end_time):
    diff = end_time - start_time
    print("Execution Time :", diff)
    return diff
```

```python
def show(df):
    display(df)
```

```python
def row_count(df):
    print("Total Rows :", df.count())
```
